```python
# ============================================
# STEP 1: IMPORT LIBRARIES
# ============================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
import pickle
import time

print("TensorFlow version:", tf.__version__)
print("NumPy version:", np.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
# ============================================
# STEP 1B: CIFAR-10 DATASET PATH
# ============================================

DATA_PATH = os.path.expanduser(
    "~/Desktop/Desktop/Semester/Sem 7/Project/College/cifar-10-batches-py"
)

print("Dataset path:")
print(DATA_PATH)

print("\nDataset exists:", os.path.exists(DATA_PATH))
```

```python
# ============================================
# STEP 2: LOAD CIFAR-10
# ============================================

def load_cifar_batch(file_path):
    with open(file_path, "rb") as f:
        data = pickle.load(f, encoding="bytes")

    images = data[b"data"]
    labels = data[b"labels"]

    return images, labels


# --------------------------------------------
# Load the 5 training batches
# --------------------------------------------

train_images = []
train_labels = []

for i in range(1, 6):

    file_path = os.path.join(
        DATA_PATH,
        f"data_batch_{i}"
    )

    images, labels = load_cifar_batch(file_path)

    train_images.append(images)
    train_labels.extend(labels)


# Combine all training batches
x_train = np.concatenate(
    train_images,
    axis=0
)

y_train = np.array(train_labels)


# --------------------------------------------
# Load test batch
# --------------------------------------------

test_path = os.path.join(
    DATA_PATH,
    "test_batch"
)

x_test, y_test = load_cifar_batch(
    test_path
)

x_test = np.array(x_test)
y_test = np.array(y_test)


# --------------------------------------------
# Check the loaded data
# --------------------------------------------

print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)

print("Test images:", x_test.shape)
print("Test labels:", y_test.shape)
```

```python
# ============================================
# STEP 3: RESHAPE CIFAR-10 IMAGES
# ============================================

x_train = x_train.reshape(
    -1, 3, 32, 32
).transpose(
    0, 2, 3, 1
)

x_test = x_test.reshape(
    -1, 3, 32, 32
).transpose(
    0, 2, 3, 1
)

print("Training images:", x_train.shape)
print("Test images:", x_test.shape)
```

```python
# ============================================
# STEP 6.1: ACTIVATION FUNCTION
# ============================================

def apply_activation(x, activation):

    if activation == "relu":
        return tf.keras.layers.ReLU()(x)

    elif activation == "elu":
        return tf.keras.layers.ELU()(x)

    elif activation == "gelu":
        return tf.keras.layers.Activation(
            tf.keras.activations.gelu
        )(x)

    elif activation == "swish":
        return tf.keras.layers.Activation(
            tf.keras.activations.swish
        )(x)

    elif activation == "tanh":
        return tf.keras.layers.Activation("tanh")(x)

    elif activation == "softsign":
        return tf.keras.layers.Activation("softsign")(x)

    elif activation == "softplus":
        return tf.keras.layers.Activation("softplus")(x)

    else:
        raise ValueError(
            f"Unknown activation: {activation}"
        )
```

```python
# ============================================
# STEP 6.2: RESIDUAL BLOCK
# ============================================

def residual_block(
    x,
    filters,
    stride=1,
    activation="relu"
):

    shortcut = x

    # ----------------------------------------
    # First 3x3 convolution
    # ----------------------------------------

    x = tf.keras.layers.Conv2D(
        filters=filters,
        kernel_size=3,
        strides=stride,
        padding="same",
        use_bias=False
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    x = apply_activation(
        x,
        activation
    )

    # ----------------------------------------
    # Second 3x3 convolution
    # ----------------------------------------

    x = tf.keras.layers.Conv2D(
        filters=filters,
        kernel_size=3,
        strides=1,
        padding="same",
        use_bias=False
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)

    # ----------------------------------------
    # Shortcut projection
    # ----------------------------------------

    if stride != 1 or shortcut.shape[-1] != filters:

        shortcut = tf.keras.layers.Conv2D(
            filters=filters,
            kernel_size=1,
            strides=stride,
            padding="same",
            use_bias=False
        )(shortcut)

        shortcut = tf.keras.layers.BatchNormalization()(shortcut)

    # ----------------------------------------
    # Add residual connection
    # ----------------------------------------

    x = tf.keras.layers.Add()([
        x,
        shortcut
    ])

    x = apply_activation(
        x,
        activation
    )

    return x
```

```python
# ============================================
# STEP 6.3: BUILD RESNET-18
# ============================================

def build_resnet18(
    input_shape=(32, 32, 3),
    num_classes=10,
    activation="relu"
):

    inputs = tf.keras.layers.Input(
        shape=input_shape
    )

    # ----------------------------------------
    # Initial 7x7 convolution
    # ----------------------------------------

    x = tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=7,
        strides=2,
        padding="same",
        use_bias=False
    )(inputs)

    x = tf.keras.layers.BatchNormalization()(x)

    x = apply_activation(
        x,
        activation
    )

    # ----------------------------------------
    # Max Pooling
    # ----------------------------------------

    x = tf.keras.layers.MaxPooling2D(
        pool_size=3,
        strides=2,
        padding="same"
    )(x)

    # ----------------------------------------
    # Group 1
    # 64 filters
    # ----------------------------------------

    x = residual_block(
        x,
        filters=64,
        stride=1,
        activation=activation
    )

    x = residual_block(
        x,
        filters=64,
        stride=1,
        activation=activation
    )

    # ----------------------------------------
    # Group 2
    # 128 filters
    # ----------------------------------------

    x = residual_block(
        x,
        filters=128,
        stride=2,
        activation=activation
    )

    x = residual_block(
        x,
        filters=128,
        stride=1,
        activation=activation
    )

    # ----------------------------------------
    # Group 3
    # 256 filters
    # ----------------------------------------

    x = residual_block(
        x,
        filters=256,
        stride=2,
        activation=activation
    )

    x = residual_block(
        x,
        filters=256,
        stride=1,
        activation=activation
    )

    # ----------------------------------------
    # Group 4
    # 512 filters
    # ----------------------------------------

    x = residual_block(
        x,
        filters=512,
        stride=2,
        activation=activation
    )

    x = residual_block(
        x,
        filters=512,
        stride=1,
        activation=activation
    )

    # ----------------------------------------
    # Global Average Pooling
    # ----------------------------------------

    x = tf.keras.layers.GlobalAveragePooling2D()(x)

    # ----------------------------------------
    # CIFAR-10 classification layer
    # ----------------------------------------

    outputs = tf.keras.layers.Dense(
        num_classes,
        activation="softmax"
    )(x)

    model = tf.keras.Model(
        inputs=inputs,
        outputs=outputs
    )

    return model
```

```python
# ============================================
# STEP 6.4: CREATE RELU RESNET-18
# ============================================

model = build_resnet18(
    input_shape=(32, 32, 3),
    num_classes=10,
    activation="relu"
)

model.summary()
```

```python
# ============================================
# STEP 7.1: PAPER'S LEARNING-RATE DECAY
# ============================================

lr_schedule = tf.keras.optimizers.schedules.InverseTimeDecay(
    initial_learning_rate=0.1,
    decay_steps=1,
    decay_rate=5e-4,
    staircase=False
)

print("Learning-rate schedule created.")
```

```python
# ============================================
# STEP 7.2: SGD OPTIMIZER
# ============================================

optimizer = tf.keras.optimizers.SGD(
    learning_rate=lr_schedule,
    momentum=0.9
)

print("SGD optimizer created.")
```

```python
# ============================================
# STEP 7.3: COMPILE MODEL
# ============================================

model.compile(
    optimizer=optimizer,
    loss=tf.keras.losses.CategoricalCrossentropy(),
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

```python
# ============================================
# STEP 26.4: CREATE OPTIMIZER
# ============================================

def create_optimizer():

    lr_schedule = tf.keras.optimizers.schedules.InverseTimeDecay(
        initial_learning_rate=LEARNING_RATE,
        decay_steps=1,
        decay_rate=DECAY,
        staircase=False
    )

    optimizer = tf.keras.optimizers.SGD(
        learning_rate=lr_schedule,
        momentum=MOMENTUM
    )

    return optimizer

print("Optimizer function created successfully.")
```

```python
# ============================================
# STEP 26.8: REUSABLE TRAINING FUNCTION
# ============================================

def train_activation_model(activation_name):

    print("\n" + "=" * 60)
    print(f"TRAINING: {activation_name.upper()}")
    print("=" * 60)

    # Build a fresh ResNet-18
    model = build_resnet18(
        input_shape=(32, 32, 3),
        num_classes=10,
        activation=activation_name
    )

    # Create a fresh optimizer
    optimizer = create_optimizer()

    # Compile
    model.compile(
        optimizer=optimizer,
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=["accuracy"]
    )

    # Train
    start_time = time.time()

    history = model.fit(
        x_train,
        y_train,
        validation_data=(x_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=1
    )

    training_time = time.time() - start_time

    # Evaluate on test set
    test_loss, test_accuracy = model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    print("\n" + "-" * 40)
    print(f"{activation_name.upper()} RESULT")
    print("-" * 40)
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
    print(f"Training Time: {training_time / 60:.2f} minutes")

    return model, history, test_accuracy, training_time
```
